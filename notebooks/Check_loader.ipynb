```python
import sqlite3
conn = sqlite3.connect('excel_data.db')
cursor = conn.cursor()
cursor.execute("SELECT filename, result_json FROM files WHERE result_json IS NOT NULL")
rows = cursor.fetchall()
for row in rows:
    print(f"File: {row[0]}")
    print(f"JSON (first 500 chars): {row[1][:500]}...")
conn.close()
```
